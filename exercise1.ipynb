```python
## question 1a

for i in range(1, 101):
    if i % 3 == 0 and i % 5 == 0:
        print("MickeyMouse")
    elif i % 3 == 0:
        print("Mickey")
    elif i % 5 == 0:
        print("Mouse")
    else:
        print(i)
```

```python
##Question 1b

results = []

for i in range(1, 101):
    if i % 15 == 0:
        results.append("MickeyMouse")
    elif i % 3 == 0:
        results.append("Mickey")
    elif i % 5 == 0:
        results.append("Mouse")
    else:
        results.append(i)

results = tuple(results)

donald_duck = tuple(
    "DonaldDuck" if x == "MickeyMouse"
    else "Donald" if x == "Mickey"
    else "Duck" if x == "Mouse"
    else x
    for x in results
)

print(donald_duck)
```

```python
## Question 2

def swap(x, y):
    temp = x
    x = y
    y = temp
    return x, y

print(swap(5, 10))
print(swap("hello", "world"))
```

```python
##Question 3

def distance(u, v):
    x1, y1 = u
    x2, y2 = v

    return ((x2 - x1)**2 + (y2 - y1)**2)**0.5

print(distance((3, 0), (0, 4)))
```

```python
##Question 4

s = """Write a program that prints the numbers from 1 to 100.
But for multiples of three print Mickey instead of the number
and for multiples of five print Mouse.
For numbers which are multiples of both three and five
print MickeyMouse"""

counts = {}

for character in s.lower():
    counts[character] = counts.get(character, 0) + 1

print(counts)

## Or if letters and characters only use this
for character in s.lower():
    if character.isalpha():
        counts[character] = counts.get(character, 0) + 1
```

```python
##Questionn 5

l = [36, 45, 58, 3, 74, 96, 64, 45, 31, 10, 24, 19, 33,
     86, 99, 18, 63, 70, 85, 85, 63, 47, 56, 42, 70, 84,
     88, 55, 20, 54, 8, 56, 51, 79, 81, 57, 37, 91, 1,
     84, 84, 36, 66, 9, 89, 50, 42, 91, 50, 95, 90, 98,
     39, 16, 82, 31, 92, 41, 45, 30, 66, 70, 34, 85,
     94, 5, 3, 36, 72, 91, 84, 34, 87, 75, 53, 51, 20,
     89, 51, 20]

def unique_numbers(l):
    unique = set(l)
    return unique, len(unique)

distinct, count_distinct = unique_numbers(l)

print("Distinct numbers:", distinct)
print("Number of distinct values:", count_distinct)
```

```python
## Question 6

def square(x):
    return x**2

def cube(x):
    return x**3

def sixth_power(x):
    return cube(square(x))

print(square(3))
print(cube(3))
print(sixth_power(3))
```

```python
##Question 8

def normalize(numbers):
    total = sum(numbers)

    if total == 0:
        raise ValueError("Cannot normalize a tuple with zero sum.")

    return tuple(x / total for x in numbers)

print(normalize((2, 3, 5)))

```
